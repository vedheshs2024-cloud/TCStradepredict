```python
for i in col:
    q25,q75=np.percentile(df[i],25),np.percentile(df[i],75)
    IQR = q75 - q25
    Threshold=IQR*1.5
    lower,upper=q25-Threshold,q75+Threshold
    outliers=[j for j in df[i] if j < lower or j > upper]
    for k in outliers:
        df=df.drop(df.index[df[i]==k],axis=0)
```

```python
from keras.models import Sequential
import tensorflow as tf
from tensorflow.keras.layers import Dense, Dropout, Conv1D, MaxPooling1D, Bidirectional, LSTM, Flatten

# CNN + BiLSTM hybrid model
model_cnn_bilstm = Sequential()

# -------- CNN FEATURE EXTRACTION ---------
model_cnn_bilstm.add(Conv1D(
    filters=64,
    kernel_size=1,
    activation='relu',
    input_shape=(1, 4)      # 1 timestep, 9 features
))

model_cnn_bilstm.add(MaxPooling1D(pool_size=1))
model_cnn_bilstm.add(Dropout(0.2))

# -------- Bidirectional LSTM ---------
model_cnn_bilstm.add(Bidirectional(LSTM(64, return_sequences=False)))
model_cnn_bilstm.add(Dropout(0.2))

# -------- Dense Layers ---------
model_cnn_bilstm.add(Dense(32, activation='relu'))
model_cnn_bilstm.add(Dropout(0.2))

# Output Layer
model_cnn_bilstm.add(Dense(1, activation='linear'))

# -------- Compile ---------
model_cnn_bilstm.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mse', tf.keras.metrics.MeanAbsolutePercentageError(), 'mae']
)

# -------- Early Stop Condition ---------
max_mae = 0.005

class StopCond(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        mae = logs.get('mae')
        if mae is not None and mae < max_mae:
            print(f"MAE threshold condition satisfied: {mae:.6f} < {max_mae}")
            self.model.stop_training = True

early_stop = StopCond()

# -------- Fit Model ---------
history = model_cnn_bilstm.fit(
    xtrain_reshaped, ytrain,
    epochs=200,
    batch_size=32,
    validation_split=0.1,
    verbose=1,
    callbacks=[early_stop]
)
```

```python
from keras.models import Sequential
import tensorflow as tf
from tensorflow.keras.layers import Dense, Dropout, Conv1D, MaxPooling1D, Bidirectional, LSTM, Flatten

# CNN + BiLSTM hybrid model
model_cnn_bilstm1 = Sequential()

# -------- CNN FEATURE EXTRACTION ---------
model_cnn_bilstm1.add(Conv1D(
    filters=64,
    kernel_size=1,
    activation='relu',
    input_shape=(1, 4)      # 1 timestep, 9 features
))

model_cnn_bilstm1.add(MaxPooling1D(pool_size=1))
model_cnn_bilstm1.add(Dropout(0.2))

# -------- Bidirectional LSTM ---------
model_cnn_bilstm1.add(Bidirectional(LSTM(64, return_sequences=False)))
model_cnn_bilstm1.add(Dropout(0.2))

# -------- Dense Layers ---------
model_cnn_bilstm1.add(Dense(32, activation='relu'))
model_cnn_bilstm1.add(Dropout(0.2))

# Output Layer
model_cnn_bilstm1.add(Dense(1, activation='linear'))

# -------- Compile ---------
model_cnn_bilstm1.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mse', tf.keras.metrics.MeanAbsolutePercentageError(), 'mae']
)

# -------- Early Stop Condition ---------
max_mae = 0.005

class StopCond(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        mae = logs.get('mae')
        if mae is not None and mae < max_mae:
            print(f"MAE threshold condition satisfied: {mae:.6f} < {max_mae}")
            self.model.stop_training = True

early_stop = StopCond()

# -------- Fit Model ---------
history1 = model_cnn_bilstm1.fit(
    xtrain_reshaped1, ytrain1,
    epochs=200,
    batch_size=32,
    validation_split=0.1,
    verbose=1,
    callbacks=[early_stop]
)
```
